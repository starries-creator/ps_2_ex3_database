# Insight 3: Have NBA Games Become More Competitive Over Time?

This analysis examines whether the proportion of close NBA regular-season games has changed over time.

A **close game** is defined as a game decided by **five points or fewer**. The analysis covers available seasons from 1980–81 to 2022–23.

```python
from pathlib import Path
import matplotlib.pyplot as plt

from nba.data_loader import connect_database
from nba.insight_3 import close_game_trend
from nba.insight_3_viz import plot_close_game_trend

DATA_PATH = Path("data/nba.sqlite")

print("Database found:", DATA_PATH.exists())
```

```python
con = connect_database(DATA_PATH)

try:
    trend = close_game_trend(con)
finally:
    con.close()

print("Number of seasons:", len(trend))
print("Total games:", trend["games"].sum())

display(trend.head())
display(trend.tail())
```

```python
fig = plot_close_game_trend(
    trend,
    output_path="figures/insight_3_close_game_rate.png"
)

plt.show()
```

## Findings

The analysis covers 47,027 regular-season games across 42 available NBA seasons between 1980–81 and 2022–23.

The proportion of games decided by five points or fewer fluctuated considerably over time. In 1980–81, 32.77% of games were close, compared with 27.56% in 2022–23. The share reached 35.21% in 1981–82, while recent seasons generally recorded values around 25–28%.

These results suggest that close games have not become consistently more common over time. The 2012–13 season is missing from the dataset and is shown as a gap in the figure.

A limitation is that the final point margin does not fully capture competitiveness throughout a game. A large comeback or a game that remains close until the final minutes may not be accurately represented by the final score alone.
